# Preprocesamiento base de datos de Pokémon

```python
import pandas as pd

df = pd.read_csv("../data/csvs/pokedex_showdown.csv", delimiter=',', encoding='utf-8')
df.head()
```

```python
# Análisis descriptivo de los datos
print("Información general del DataFrame:")
print(df.info())
print("\nEstadísticas descriptivas:")
print(df.describe())
print("\nPrimeras filas:")
print(df.head())
print("\nColumnas del DataFrame:")
print(df.columns.tolist())
print("\nTipos de datos:")
print(df.dtypes)
```

```python
print("Nombres de las columnas:")
print(df.columns.tolist())
```

```python
# Eliminar columnas irrelevantes
df.drop(columns=['internal_id', 'pokedex_number', 'egg_groups', 'base_species', 'forme', 'color'], inplace=True)
print("Columnas eliminadas. Nuevas columnas:")
print(df.columns.tolist())
```

```python
import pandas as pd

# Columnas que vamos a rellenar desde el base

cols_to_fill = [
    'types',
    'hp', 'attack', 'defense',
    'sp_attack', 'sp_defense', 'speed',
    'height_m', 'weight_kg',
    'tier'
]

# Crear nombre base del Pokémon

df['base_name'] = df['name'].str.split('-').str[0]

# Definir qué consideramos valor vacío

def is_empty(x):
    return pd.isna(x) or x == '' or x == 'None'

# Función para heredar valores del Pokémon base

def fill_from_base(df, column):
    # Mapa base_name -> valor válido
    base_map = (
        df[~df[column].apply(is_empty)]
        .set_index('base_name')[column]
        .to_dict()
    )

    # Rellenar solo si está vacío
    df[column] = df.apply(
        lambda row: base_map.get(row['base_name'], row[column])
        if is_empty(row[column])
        else row[column],
        axis=1
    )

# Aplicar a todas las columnas a rellenar
for col in cols_to_fill:
    fill_from_base(df, col)

# Limpiar columna temporal

df.drop(columns=['base_name'], inplace=True)

# Comprobación final

print('Valores vacíos restantes por columna:')
print(df[cols_to_fill].apply(lambda col: col.map(is_empty)).sum())
```

```python
# Filtrar por tiers específicos
tiers = ['OU', 'UUBL', 'UU', 'RUBL', 'RU', 'NUBL', 'NU', 'PUBL', 'PU', 'ZUBL', 'ZU', 'LC', 'NFE', 'Uber','Illegal']
df = df[df['tier'].isin(tiers)]
print(f"DataFrame filtrado. Número de filas: {len(df)}")
print("Tiers únicos restantes:")
print(df['tier'].unique())
df.drop(columns=['tier'], inplace=True)
```

```python
# Dividir la columna 'type' en 'type1' y 'type2'
df[['type1', 'type2']] = df['types'].str.split(',', expand=True)
print("Columna 'type' dividida en 'type1' y 'type2'.")
print("Ejemplos:")
print(df[['types', 'type1', 'type2']].head())
```

```python
# Rellenar type2 con type1 donde type2 es None
df['type2'] = df['type2'].fillna(df['type1'])
print("Type2 rellenado con type1 donde era None.")
print("Ejemplos:")
print(df[['types', 'type1', 'type2']].head())
```

```python
# Eliminar la columna 'types'
df.drop(columns=['types'], inplace=True)
print("Columna 'types' eliminada.")
print("Columnas restantes:")
print(df.columns.tolist())
```

```python
# Dividir la columna 'abilities' en 'ability1', 'ability2', 'ability3'
split_abilities = df['abilities'].str.split(',', expand=True)
df['ability1'] = split_abilities[0]
df['ability2'] = split_abilities[1]
df['ability3'] = split_abilities[2]
print("Columna 'abilities' dividida en 'ability1', 'ability2', 'ability3'.")
print("Ejemplos:")
print(df[['abilities', 'ability1', 'ability2', 'ability3']].head())
```

```python
# Eliminar la columna 'abilities'
df.drop(columns=['abilities'], inplace=True)
print("Columna 'abilities' eliminada.")
print("Columnas restantes:")
print(df.columns.tolist())
```

```python
# Ver valores faltantes
print("Valores faltantes por columna:")
print(df.isnull().sum())
print("\nPorcentaje de valores faltantes:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
# Eliminar la columna 'generation'
df.drop(columns=['generation'], inplace=True)
print("Columna 'generation' eliminada.")
print("Columnas restantes:")
print(df.columns.tolist())
```
