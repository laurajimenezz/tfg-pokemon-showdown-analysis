Calculamos la suma de la vida total de todos los Pokemon de cada Jugador en cada turno.

Primero sustituimos los valores faltantes en Vida_J1 y Vida_J2 por la vida que apaereca que tenga ese Pokémon en un turno anterior y si no ha aparecido antes en la batalla sustituir por el valor inicial que es 100.

```python
import pandas as pd

# Ordenamos el DataFrame por batalla y turno para mantener el orden temporal
df_filtrado = df_filtrado.sort_values(['Id_Batalla', 'Turno']).reset_index(drop=True)

# Diccionarios para guardar la última vida conocida por Pokémon dentro de cada batalla
ultima_vida_j1 = {}
ultima_vida_j2 = {}

# Listas para las nuevas vidas corregidas
vidas_corr_j1 = []
vidas_corr_j2 = []

# Recorremos las filas en orden
for idx, fila in df_filtrado.iterrows():
    battle = fila['Id_Batalla']
    activo_j1 = fila['Activo_Jugador1']
    activo_j2 = fila['Activo_Jugador2']
    vida_j1 = fila['Vida_J1']
    vida_j2 = fila['Vida_J2']

    # Inicializamos diccionarios por batalla
    if battle not in ultima_vida_j1:
        ultima_vida_j1[battle] = {}
        ultima_vida_j2[battle] = {}

    # Jugador 1
    if pd.isna(vida_j1) or vida_j1 == '':
        # Si el Pokémon ya apareció antes, usamos su última vida
        if activo_j1 in ultima_vida_j1[battle]:
            vida_j1 = ultima_vida_j1[battle][activo_j1]
        else:
            vida_j1 = 100  # Si no ha aparecido, vida inicial = 100
    else:
        # Guardamos la vida actual como la última conocida
        ultima_vida_j1[battle][activo_j1] = vida_j1

    # Jugador 2
    if pd.isna(vida_j2) or vida_j2 == '':
        if activo_j2 in ultima_vida_j2[battle]:
            vida_j2 = ultima_vida_j2[battle][activo_j2]
        else:
            vida_j2 = 100
    else:
        ultima_vida_j2[battle][activo_j2] = vida_j2

    # Guardamos las vidas corregidas
    vidas_corr_j1.append(vida_j1)
    vidas_corr_j2.append(vida_j2)

# Asignamos las vidas corregidas al DataFrame
df_filtrado['Vida_J1'] = vidas_corr_j1
df_filtrado['Vida_J2'] = vidas_corr_j2

print(df_filtrado[['Id_Batalla', 'Turno', 'Activo_Jugador1', 'Vida_J1', 'Activo_Jugador2', 'Vida_J2']].head(20))
```

```python
# Columnas de Pokémon de cada jugador
pokemon_cols_j1 = [f'Pokemon{i}_Jugador1' for i in range(1, 7)]
pokemon_cols_j2 = [f'Pokemon{i}_Jugador2' for i in range(1, 7)]

# Número de Pokémon por jugador (contando solo los que no son NaN)
df_filtrado['Num_Pokemon_J1'] = df_filtrado[pokemon_cols_j1].notna().sum(axis=1)
df_filtrado['Num_Pokemon_J2'] = df_filtrado[pokemon_cols_j2].notna().sum(axis=1)

# Primero obtenemos las columnas de los Pokémon
pokemon_cols_j1 = [f'Pokemon{i}_Jugador1' for i in range(1, 7)]
pokemon_cols_j2 = [f'Pokemon{i}_Jugador2' for i in range(1, 7)]

# Inicializamos las columnas de vida de cada Pokémon con 100
for col in pokemon_cols_j1:
    df_filtrado[f'Vida_{col}'] = 100
for col in pokemon_cols_j2:
    df_filtrado[f'Vida_{col}'] = 100

# Ordenamos por Id_Batalla y Turno para procesar correctamente
df_filtrado = df_filtrado.sort_values(['Id_Batalla', 'Turno'])

# Diccionarios para llevar la vida actual de cada Pokémon por batalla
vida_actual_j1 = {}
vida_actual_j2 = {}

# Listas para almacenar la suma de vida total por fila
suma_vida_j1 = []
suma_vida_j2 = []

# Procesamos fila por fila
for idx, fila in df_filtrado.iterrows():
    battle = fila['Id_Batalla']

    # Inicializamos si es la primera vez que vemos esta batalla
    if battle not in vida_actual_j1:
        vida_actual_j1[battle] = [100]*6
        vida_actual_j2[battle] = [100]*6

    # Pokémon activo de cada jugador (si hay)
    activo_j1 = fila['Activo_Jugador1']
    activo_j2 = fila['Activo_Jugador2']

    # Actualizamos la vida del Pokémon activo con la vida de ese turno
    if pd.notna(activo_j1) and pd.notna(fila['Vida_J1']):
        # Buscamos la posición del Pokémon activo en su equipo
        try:
            pos = fila[pokemon_cols_j1].tolist().index(activo_j1)
            vida_actual_j1[battle][pos] = fila['Vida_J1']
        except ValueError:
            pass  # Pokémon no encontrado, ignoramos
    if pd.notna(activo_j2) and pd.notna(fila['Vida_J2']):
        try:
            pos = fila[pokemon_cols_j2].tolist().index(activo_j2)
            vida_actual_j2[battle][pos] = fila['Vida_J2']
        except ValueError:
            pass

    # Calculamos la suma de vida de los Pokémon existentes (no NaN)
    num_j1 = fila['Num_Pokemon_J1']
    num_j2 = fila['Num_Pokemon_J2']

    suma_vida_j1.append(sum(vida_actual_j1[battle][:num_j1]))
    suma_vida_j2.append(sum(vida_actual_j2[battle][:num_j2]))

# Creamos las nuevas columnas en el DataFrame
df_filtrado['Suma_Vida_J1'] = suma_vida_j1
df_filtrado['Suma_Vida_J2'] = suma_vida_j2
# Columnas de vida de cada Pokémon de cada jugador
vida_cols_j1 = [f'Vida_Pokemon{i}_Jugador1' for i in range(1, 7)]
vida_cols_j2 = [f'Vida_Pokemon{i}_Jugador2' for i in range(1, 7)]

# Eliminamos estas columnas del DataFrame
df_filtrado = df_filtrado.drop(columns=vida_cols_j1 + vida_cols_j2, errors='ignore')


# Comprobamos resultado
print(df_filtrado[['Id_Batalla', 'Turno','Num_Pokemon_J1','Activo_Jugador1','Vida_J1', 'Suma_Vida_J1', 'Num_Pokemon_J2','Activo_Jugador2','Vida_J2', 'Suma_Vida_J2']].head(20))
```

Ahora vamos a generar otro archivo CSV donde vamos a adjuntar características de los Pokémon a la base de datos de las batallas

```python
pokedex = pd.read_csv("../data/csvs/Pokemon_filtrado.csv", delimiter=',', encoding='utf-8')
pokedex.head()
batallas=df_filtrado.copy()
```

```python
def pokemon_base(nombre):
    if pd.isna(nombre):
        return nombre

    especiales = {
        'Milotic-Sevii': 'Milotic'
    }

    if nombre in especiales:
        return especiales[nombre]

    return nombre.split('-')[0]

def merge_pokemon_activo(
    batallas,
    pokedex,
    col_activo,
    sufijo
):
    """
    Une información del Pokédex al Pokémon activo de un jugador.
    Corrige nombres SOLO cuando no hay match directo.
    """

    # Intento de merge directo
    merge_directo = batallas.merge(
        pokedex,
        how='left',
        left_on=col_activo,
        right_on='nombre'
    )

    # Detectar los que no hacen match
    mask_no_match = merge_directo['tipo_1'].isna()

    # Crear columna corregida
    col_fix = f'{col_activo}_fix'
    batallas[col_fix] = batallas[col_activo]

    # Corregir SOLO los problemáticos
    batallas.loc[mask_no_match, col_fix] = (
        batallas.loc[mask_no_match, col_activo]
        .apply(pokemon_base)
    )

    # Merge final
    batallas = batallas.merge(
        pokedex[
            [
                'nombre',
                'tipo_1',
                'tipo_2',
                'velocidad',
                'ataque',
                'ataque_especial',
                'defensa',
                'defensa_especial'
            ]
        ],
        how='left',
        left_on=col_fix,
        right_on='nombre',
        suffixes=('', f'_{sufijo}')
    )

    # Renombrar columnas añadidas
    batallas = batallas.rename(
        columns={
            'tipo_1': f'tipo_1_activo_{sufijo}',
            'tipo_2': f'tipo_2_activo_{sufijo}',
            'velocidad': f'velocidad_activo_{sufijo}',
            'ataque': f'ataque_activo_{sufijo}',
            'ataque_especial': f'ataque_esp_activo_{sufijo}',
            'defensa': f'defensa_activo_{sufijo}',
            'defensa_especial': f'defensa_esp_activo_{sufijo}'
        }
    )

    # Limpiar columnas auxiliares
    batallas = batallas.drop(columns=['nombre'])

    return batallas

batallas = merge_pokemon_activo(
    batallas,
    pokedex,
    col_activo='Activo_Jugador1',
    sufijo='j1'
)

batallas = merge_pokemon_activo(
    batallas,
    pokedex,
    col_activo='Activo_Jugador2',
    sufijo='j2'
)
# Verificar que no hay NaN en tipos
batallas[
    [
        'tipo_1_activo_j1',
        'tipo_1_activo_j2'
    ]
].isna().sum()
```

```python
batallas = batallas.drop(
    columns=['Activo_Jugador1_fix', 'Activo_Jugador2_fix'],
    errors='ignore'
)

# Guardar el DataFrame en un archivo CSV
ruta_salida = "../data/csvs/batallas_detallado.csv"
batallas.to_csv(ruta_salida, index=False, encoding='utf-8')
```
